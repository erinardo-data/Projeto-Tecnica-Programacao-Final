```python
import pandas as pd

# Substitua pela URL "Raw" do seu arquivo no GitHub
url = 'AirbnbRJRentals.csv'

# Se o arquivo estiver na mesma pasta no Codespaces, use apenas pd.read_csv('AirbnbRJRentals.csv')
df = pd.read_csv(url)
```

```python
# Substituir os valores vazios em 'reviews_per_month' por 0
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# Verificar quantos valores nulos ainda restam no dataset
print(df.isnull().sum())
```

```python
# Converter a coluna para o formato data (valores vazios se tornarão NaT - Not a Time)
df['last_review'] = pd.to_datetime(df['last_review'])

print(df['last_review'].dtypes)
```

```python
# Filtrar imóveis com preços maiores que zero e menores ou iguais a R$ 5.000
df_limpo = df[(df['price'] > 0) & (df['price'] <= 5000)]

# Filtrar exigência de noites mínimas para no máximo 365 dias (1 ano)
df_limpo = df_limpo[df_limpo['minimum_nights'] <= 365]

# Comparar o tamanho da tabela antes e depois da limpeza
print(f"Linhas antes da limpeza: {df.shape[0]}")
print(f"Linhas após a limpeza: {df_limpo.shape[0]}")
```
